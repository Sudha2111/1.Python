```python
#A program processes a list of numbers and needs to find the largest value.
# Write logic to identify and return the largest number from a given list.
#Function definition → Function call → Initialize → Loop → Compare → Update → Return → Print

def Largest (Large):
    number=Large[0]
    for num in Large:
       if num > number:
         number=num
       return number
        
Large =[98,77,64,35,22]
print(Largest(Large))



```

```python
#A calendar system verifies whether a given year is a leap year based on standard leap year rules.

yer=int(input("Enter year:"))
if yer % 400==0 or (yer %4 ==0 and yer % 100!=0):
   print("It is a Leap Year",yer)
else:
    print("It is  NOT a Leap Year",yer)
```
